```python
import pandas as pd
data_frame = pd.read_csv('data.csv')
data_frame.shape
```

```python
data_frame.describe()
```

```python
data_frame.values
```

```python
data_frame[data_frame["Age"]>40].head()
```

```python
df1 = pd.DataFrame(data_frame, columns=['Name', 'Wage', 'Value'])
```

```python
df1['Wage'] = df1['Wage'].str.replace('€', '')
df1['Value'] = df1['Value'].str.replace('€', '')
df1
```

```python
df1['Wage'] = df1['Wage'].replace({'K': '*1e3', 'M': '*1e6'}, regex=True).map(pd.eval).astype(int)
df1['Value'] = df1['Value'].replace({'K': '*1e3', 'M': '*1e6'}, regex=True).map(pd.eval).astype(int)
```

```python
df1
```

```python
df1['difference'] = df1['Value'] - df1['Wage']
df1.sort_values('difference', ascending=False)
```

```python
import seaborn as sns
sns.set()

graph = sns.scatterplot(x='Wage', y='Value', data=df1)
graph
```

```python
from bokeh.plotting import figure,show
from bokeh.models import HoverTool

TOOLTIPS = HoverTool(tooltips=[('index', '$index'), 
                               ('(Wage,Value)', '(@Wage, @Value)'), 
                               ('Name', '@Name')]
                    )


p = figure( width=700, height=700, title='soccer 2019', x_axis_label='Wage', y_axis_label='Value', tools=[TOOLTIPS])
p.circle('Wage', 'Value', size=10, source=df1)
show(p)
```
